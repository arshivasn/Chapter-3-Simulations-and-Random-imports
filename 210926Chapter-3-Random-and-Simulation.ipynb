```python
def getSuit(card):
    if card < 13:
        return "Diamond"
    if card < 26:
        return "Club"
    if card < 39:
        return "Heart"
    return "Spade"

def getRank(card):
    rank = card % 13 + 2
    if rank == 11:
        return 'J'
    if rank == 12:
        return 'Q'
    if rank == 13:
        return 'K'
    if rank == 14:
        return 'A'
    return str(rank) 
    
def isFlush(hand):
    suit0 = getSuit(hand[0])

    for card in hand:
        if getSuit(card) != suit0:
            return False
    return True 
```

```python
#simple simulation --> List slicing 
#var[start_index:end_index]
hand = deck[0:5]
print(hand)

#had a function to vall for checking if cards form a flush then we could execute this -->
print(isFlush(hand))
```

```python
for i in range(1000):
    random.shuffle(deck)
    hand = deck[0:5]

    if isFlush(hand):
        print("We have got one!")
```

```python
#counting 
flush = 0
for i in range(1000000):
    random.shuffle(deck)
    hand = deck[0:5]

    if isFlush(hand):
        flush = flush + 1
print("probability of a flush:", flush/1000000) #probability of a flush in these many events 
```

```python
import random
N = 10000
countList1 = [0] * 51 #list of 51 counters, initialised to 0; in worst case, it takes 50 dice rolls to end the game :)

for i in range(N):
    piece = 0
    rolls = 0
    
    while piece < 50:
        piece = piece + random.randint(1, 6)
        rolls = rolls + 1 
    #print(piece, end= " ")
    countList1[rolls] = countList1[rolls] + 1 #number of times game ends in 'n' rolls
print(countList1)
```

```python
def specialRoute(piece):
    if piece == 6:
        return 14
    if piece == 16:
        return 4
    if piece == 17:
        return 23
    if piece == 27:
        return 33
    if piece == 29:
        return 10
    if piece == 38:
        return 43
    if piece == 39:
        return 20
    if piece == 45:
        return 34
    return piece
    
```

```python
import random
N = 10000
resultList = []
for i in range(N):

    piece = 0
    rolls = 0

    while piece < 50:
        piece = piece + random.randint(1, 6)
        piece = specialRoute(piece)
        rolls = rolls + 1
    resultList.append(rolls)
```

```python
countList2 = [0] * (max(resultList) + 1)

for result in resultList:
    countList2[result] = countList2[result] + 1

# importing the modules
import matplotlib.pyplot as plt

# data to be plotted
x = range(5, 51)
y1 = countList1[5:51]
y2 = countList2[5:51]

# plotting
plt.title("Rolls needed to End the Game")
plt.xlabel("Number of Rolls")
plt.ylabel("Out of " + str(N) + " times")
plt.plot(x, y1, label='Without S&L')
plt.plot(x, y2, label='With S&L')
plt.legend()
plt.show()
```
